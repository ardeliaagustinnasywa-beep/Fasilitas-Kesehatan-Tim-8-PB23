```python
# Memilih fitur untuk clustering sesuai data fasilitas kesehatan
features = [
    'JUMLAH RUMAH SAKIT',
    'JUMLAH PUSKESMAS',
    'JUMLAH POSYANDU',
    'JUMLAH APOTEK',
    'JUMLAH KLINIK'
]
```

```python
# Standardisasi data sesuai kolom fasilitas kesehatan
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df[
    ['JUMLAH RUMAH SAKIT',
     'JUMLAH PUSKESMAS',
     'JUMLAH POSYANDU',
     'JUMLAH APOTEK',
     'JUMLAH KLINIK']
])
```

```python
# Menentukan jumlah cluster menggunakan metode Elbow
inertia = []
k_values = range(1, 11)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto')
    kmeans.fit(scaled_features)   # <-- sudah memakai fasilitas kesehatan
    inertia.append(kmeans.inertia_)
```

```python
# Plot Elbow Method menggunakan data fasilitas kesehatan
plt.figure(figsize=(8, 5))
plt.plot(k_values, inertia, marker='o')
plt.title("Elbow Method untuk Menentukan Jumlah Cluster Optimal (Data Fasilitas Kesehatan)")
plt.xlabel("Jumlah Cluster (k)")
plt.ylabel("Inertia")
plt.grid()
plt.show()
```

```python
# Menggunakan jumlah cluster optimal
optimal_k = 4 # Berdasarkan elbow method
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init='auto')
df['cluster'] = kmeans.fit_predict(scaled_features)
```

```python
# Mengurangi dimensi data untuk visualisasi 
pca = PCA(n_components=2)
reduced_features = pca.fit_transform(scaled_features)
```

```python
colors = ['red', 'blue', 'green', 'purple']
```

```python
# Pastikan cluster disimpan 
df['cluster'] = kmeans.fit_predict(scaled_features)

# Plot PCA clustering (versi fasilitas kesehatan)
plt.figure(figsize=(10, 7))

for cluster in range(optimal_k):
    cluster_points = reduced_features[df['cluster'] == cluster]

    plt.scatter(
        cluster_points[:, 0],
        cluster_points[:, 1],
        s=50,
        label=f'Cluster {cluster}',
        color=colors[cluster]
    )

# Hitung posisi centroid dalam ruang PCA
centroids_reduced = pca.transform(kmeans.cluster_centers_)

plt.scatter(
    centroids_reduced[:, 0],
    centroids_reduced[:, 1],
    s=200,
    c='black',
    marker='X',
    label='Centroids'
)

plt.title("Clustering K-Means Fasilitas Kesehatan (PCA 2D)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid()
plt.legend()
plt.show()
```
